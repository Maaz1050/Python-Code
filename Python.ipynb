```python
#method overriding
class A:
  def show(self):
    self.n1 = int(input())

class B:
  def show(self):
    super().show()
    self.n2 = int(input())

class C(B,A):
  def show(self):
    super().show()
    self.n3 = self.n1+self.n2
    print("Addition is : ",self.n3)

c=C()
c.show()
```

```python
# method overriding in multiple classes
class A:
  def show(self):
    self.n1 = int(input())

class B:
  def show(self):
    self.n2 = int(input())

class C(B,A):
  def show(self):
    A.show(self)   # class name . function name
    B.show(self)
    self.n3 = self.n1+self.n2
    print("Addition is : ",self.n3)

c=C()
c.show()
```

```python
#encapsulation = wrapping / binding data into single unit
# access modifier : public, private and protected

class Sample:
  def __init__(self):
    self.name = "Afraaz Shahid Momin" # public
    self._age = 19  # protected (_)
    self.__balance = 10000000000000000 # private(__)

s=Sample()
print(s.name)
print(s._age)   # still accessible
print(s._Sample__balance)  # private value is accessible by _classname
```

```python
# setter and getter method
class Sample:
  def __init__(self):
    self.name = "Afraaz Shahid Momin" # public
    self._age = 19  # protected (_)
    self.__balance = 10000000000000000 # private(__)

  def set_data(self,ammount): #setter
    self.__balance = ammount

  def get_data(self): #getter
    print("balance : ",self.__balance)

s=Sample()
s.set_data(999999999999999)
s.get_data()
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def calculate_result(self):
        total = sum(self.marks)
        average = total / 3

        print("Total:", total)
        print("Average: ",average)

        if average >= 40:
            print("Result: Pass")
        else:
            print("Result: Fail")


name = input()
marks = list(map(int, input().split()))

student = Student(name, marks)
student.calculate_result()
```

```python
class Theory:
    def __init__(self, theory_marks):
        self.theory_marks = theory_marks


class Practical:
    def __init__(self, practical_marks):
        self.practical_marks = practical_marks


class Result(Theory, Practical):
    def __init__(self, theory_marks, practical_marks):
        Theory.__init__(self, theory_marks)
        Practical.__init__(self, practical_marks)

    def total_marks(self):
        return self.theory_marks + self.practical_marks


theory = int(input())
practical = int(input())

result = Result(theory, practical)
print(result.total_marks())
```
